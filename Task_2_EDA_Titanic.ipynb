The objective of this task is to understand the dataset using descriptive statistics and visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
print("libraries imported successfully")
```

```python
df=pd.read_csv("Titanic-Dataset.csv")
print("dataset loaded successfully")
```

```python
df.head()
```

```python
print("no. of colums:",df.shape[1])
print("no. of rows:",df.shape[0])
print(df.columns.tolist())
df.dtypes
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()
print(missing_values)
```

```python
plt.figure(figsize=(9,4))
sns.heatmap(df.isnull(),cbar=False)
plt.title("missing values")
plt.show()
```

```python
df.describe()
```

```python
numeric_col=["Age","Fare","SibSp","Parch"]
summary_statistics=pd.DataFrame({"Mean":df[numeric_col].mean(),
                                 "Median":df[numeric_col].median(),
                                 "Std_dev":df[numeric_col].std()})
print(summary_statistics)
```

```python
summary=df[numeric_col].describe()
print(summary)
```

```python
#histogram of age
plt.figure(figsize=(9,4))
sns.histplot(df["Age"],bins=30,kde=True)
plt.title("Age distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
#histogram of fare
plt.figure(figsize=(9,4))
sns.histplot(df["Fare"],bins=30,kde=True)
plt.title("Passenger fare distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
#boxplot of age
plt.figure(figsize=(9,4))
sns.boxplot(x=df["Age"])
plt.title("Age distribution")
plt.xlabel("Age")
plt.show()
```

```python
#boxplot of fare
plt.figure(figsize=(9,4))
sns.boxplot(x=df["Fare"])
plt.title("passenger Fare distribution")
plt.xlabel("Fare")
plt.show()
```

```python
# gender count plot
plt.figure(figsize=(9,5))
sns.countplot(x="Sex",data=df)
plt.title("passenger gender distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# survival count plot
plt.figure(figsize=(9,5))
sns.countplot(x="Survived",data=df)
plt.title("passenger survived distribution")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
survival_labels = df["Survived"].map({
    0: "Did Not Survive",
    1: "Survived"
})

plt.figure(figsize=(7, 5))

sns.countplot(x=survival_labels)

plt.title("Passenger Survival Distribution")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="Sex",
              hue="Survived",
              data=df
)
plt.title("gender survival distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show
```

```python
#passenger class v/s survival
plt.figure(figsize=(8,5))
sns.countplot(
    x="Pclass",
    hue="Survived",
    data=df
    )
plt.title("passenger class v/s survival")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100
print(survival_by_class)
```

```python
survival_by_class.plot(kind="bar", figsize=(8, 5))

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")

plt.show()
```

```python
survival_by_gender = df.groupby("Sex")["Survived"].mean() * 100

print(survival_by_gender)
```

```python
survival_by_gender.plot(kind="bar", figsize=(8, 5))

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Survived",
    y="Age",
    data=df
)

plt.title("Age Distribution by Survival Status")
plt.xlabel("Survived")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="Survived",
    y="Fare",
    data=df
)

plt.title("Fare Distribution by Survival Status")
plt.xlabel("Survived")
plt.ylabel("Fare")

plt.show()
```

```python
numeric_data = df[
    ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]
]
```

```python
correlation_matrix = numeric_data.corr()

correlation_matrix
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Features")

plt.show()
```

```python
skewness = df[numeric_col].skew()

print(skewness)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Fare"], bins=30, kde=True)

plt.title("Fare Distribution and Skewness")
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.show()
```

```python
fig = px.scatter(
    df,
    x="Age",
    y="Fare",
    color="Survived",
    title="Age vs Fare by Survival Status",
    hover_data=["Sex", "Pclass"]
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x="Age",
    color="Survived",
    nbins=30,
    title="Age Distribution by Survival Status"
)

fig.show()
```

```python
eda_summary = pd.DataFrame({
    "Feature": ["Age", "Fare", "Pclass", "SibSp", "Parch"],
    "Mean": [
        df["Age"].mean(),
        df["Fare"].mean(),
        df["Pclass"].mean(),
        df["SibSp"].mean(),
        df["Parch"].mean()
    ],
    "Median": [
        df["Age"].median(),
        df["Fare"].median(),
        df["Pclass"].median(),
        df["SibSp"].median(),
        df["Parch"].median()
    ],
    "Std Dev": [
        df["Age"].std(),
        df["Fare"].std(),
        df["Pclass"].std(),
        df["SibSp"].std(),
        df["Parch"].std()
    ]
})

eda_summary
```

```python
eda_summary.to_csv("eda_summary.csv", index=False)
print("EDA summary saved successfully!")
```
